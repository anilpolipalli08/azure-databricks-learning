```python
import websocket
import json

ws = websocket.create_connection("wss://ws.kraken.com/v2")

subscribe_message = {
    "method": "subscribe",
    "params": {
        "channel": "trade",
        "symbol": ["BTC/USD"]
    }
}

ws.send(json.dumps(subscribe_message))

for i in range(10):
    message = ws.recv()
    print(message)

ws.close()
```
