```python
data = [
    (1, "Anil", "Bangalore"),
    (2, "Ravi", "Hyderabad"),
    (3, "Kiran", "Chennai")
]

columns = ["id", "name", "city"]

df = spark.createDataFrame(data, columns)

df.write.format("delta").mode("overwrite").saveAsTable("customers")
```

```python
spark.sql("DESCRIBE DETAIL customers").select("name", "format", "location").show(truncate=False)
```
